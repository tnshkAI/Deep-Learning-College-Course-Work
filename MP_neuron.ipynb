# Implementation of 16 possible boolean functions using McCulloch-Pitts Neuron

### Naming/Numbering used in this program for different functions:
        "FALSE"        : 1,
        "AND"          : 2,
        "x1 AND NOT x2": 3,
        "x1"           : 4,
        "NOT x1 AND x2": 5,
        "x2"           : 6,
        "XOR"          : 7,
        "OR"           : 8,
        "NOR"          : 9,
        "XNOR"         : 10,
        "NOT x2"       : 11,
        "x1 OR NOT x2" : 12,
        "NOT x1"       : 13,
        "NOT x1 OR x2" : 14,
        "NAND"         : 15,
        "TRUE"         : 16  

```python
class MPNeuron:
    def __init__(self, weights, threshold):
        self.weights = weights
        self.threshold = threshold

    def activate(self, x):
        weighted_sum = sum(w * xi for w, xi in zip(self.weights, x))
        return 1 if weighted_sum >= self.threshold else 0
```

```python
neuron_1 = MPNeuron([0, 0], 1)       # FALSE       0000
neuron_2 = MPNeuron([1, 1], 2)       # AND         0001
neuron_3 = MPNeuron([1, -1], 1)      # x1 AND !x2 0010
neuron_4 = MPNeuron([1, 0], 1)       # x1          0011
neuron_5 = MPNeuron([-1, 1], 1)      # !x1 AND x2 0100
neuron_6 = MPNeuron([0, 1], 1)       # x2          0101

# neuron_7 = MPNeuron(...)           # XOR         0110
# Cannot be implemented by a single MP neuron

neuron_8 = MPNeuron([1, 1], 1)       # OR          0111
neuron_9 = MPNeuron([-1, -1], 0)     # NOR         1000

# neuron_10 = MPNeuron(...)          # XNOR        1001
# Cannot be implemented by a single MP neuron

neuron_11 = MPNeuron([0, -1], 0)     # !x2         1010
neuron_12 = MPNeuron([1, -1], 0)     # x1 OR !x2   1011
neuron_13 = MPNeuron([-1, 0], 0)     # !x1         1100
neuron_14 = MPNeuron([-1, 1], 0)     # !x1 OR x2   1101
neuron_15 = MPNeuron([-1, -1], -1)   # NAND        1110
neuron_16 = MPNeuron([0, 0], 0)      # TRUE        1111
```

```python
inputs = [
    [0,0], [0,1], [1,0], [1,1]
]

neurons = {
    neuron_1 : 1, 
    neuron_2 : 2,
    neuron_3 : 3, 
    neuron_4 : 4, 
    neuron_5 : 5, 
    neuron_6 : 6,
    neuron_8 : 8,
    neuron_9 : 9,
    neuron_11 : 11, 
    neuron_12 : 12,
    neuron_13 : 13, 
    neuron_14 : 14,
    neuron_15 : 15, 
    neuron_16 : 16
}
```

```python
for neuron, index in neurons.items():
    print(f"Neuron: {index}")
    for input in inputs:
        print(f"{input}  {neuron.activate(input)}")
    print("*" * 10)
```

```python
def XOR(x1, x2):
    or_neuron   = MPNeuron([1, 1], 1)
    nand_neuron = MPNeuron([-1, -1], -1)
    output_neuron  = MPNeuron([1, 1], 2)

    # Hidden layer
    h1 = or_neuron.activate([x1, x2])
    h2 = nand_neuron.activate([x1, x2])

    # Output layer
    return output_neuron.activate([h1, h2])

inputs = [(0,0), (0,1), (1,0), (1,1)]

for x in inputs:
    print(x, XOR(x[0], x[1]))
```

```python
def XNOR(x1, x2):
    and_neuron = MPNeuron([1, 1], 2)
    nor_neuron = MPNeuron([-1, -1], 0)
    output_neuron = MPNeuron([1, 1], 1)

    # Hidden layer
    h1 = and_neuron.activate([x1, x2])
    h2 = nor_neuron.activate([x1, x2])

    # Output layer
    return output_neuron.activate([h1, h2])

for x in inputs:
    print(x, XNOR(x[0], x[1]))
```
