# Implementation of 16 possible boolean functions using Perceptron

```python
class Perceptron:
    def __init__(self, weights, bias):
        self.weights = weights
        self.bias = bias

    def predict(self, x):
        z = sum(w * xi for w, xi in zip(self.weights, x)) + self.bias
        return 1 if z >= 0 else 0
```

```python
from itertools import product

inputs = [
    (0, 0),
    (0, 1),
    (1, 0),
    (1, 1)
]


def train_perceptron(targets):
    """
    Find weights w1, w2 and bias b
    that implement the Boolean function.
    """

    # Search over a reasonable range
    values = [i / 2 for i in range(-6, 7)]

    for w1 in values:
        for w2 in values:
            for b in values:

                correct = True

                for x, target in zip(inputs, targets):

                    z = w1*x[0] + w2*x[1] + b
                    prediction = 1 if z >= 0 else 0

                    if prediction != target:
                        correct = False
                        break

                if correct:
                    return [w1, w2], b

    return None
```

```python
functions = {
    "FALSE"  : [0, 0, 0, 0],
    "AND"    : [0, 0, 0, 1],
    "x1 AND NOT x2": [0, 0, 1, 0],
    "x1"     : [0, 0, 1, 1],
    "NOT x1 AND x2": [0, 1, 0, 0],
    "x2"     : [0, 1, 0, 1],
    "XOR"    : [0, 1, 1, 0],
    "OR"     : [0, 1, 1, 1],
    "NOR"    : [1, 0, 0, 0],
    "XNOR"   : [1, 0, 0, 1],
    "NOT x2" : [1, 0, 1, 0],
    "x1 OR NOT x2": [1, 0, 1, 1],
    "NOT x1" : [1, 1, 0, 0],
    "NOT x1 OR x2": [1, 1, 0, 1],
    "NAND"   : [1, 1, 1, 0],
    "TRUE"   : [1, 1, 1, 1]
}
```

```python
for name, targets in functions.items():

    result = train_perceptron(targets)

    if result is None:
        print(f"{name:20} -> Cannot be implemented by single perceptron")
        print("*" * 10)
    else:
        weights, bias = result
        print(
            f"{name} "
            # f"w = {weights}, b = {bias}"
        )
        p = Perceptron(weights, bias)
        for input in inputs:
            print(f"{input} {p.predict(input)}")
        print("*" * 10)
```

```python
xor_or = Perceptron([1, 1], -1)
xor_nand = Perceptron([-1, -1], 1)
xor_output = Perceptron([1, 1], -2)

def XOR(x1, x2):

    h1 = xor_or.predict([x1, x2])
    h2 = xor_nand.predict([x1, x2])

    return xor_output.predict([h1, h2])


inputs = [(0,0), (0,1), (1,0), (1,1)]

print("XOR:")
for x1, x2 in inputs:
    print(x1, x2, "->", XOR(x1, x2))
```

```python
xnor_and = Perceptron([1, 1], -2)
xnor_nor = Perceptron([-1, -1], 0)
xnor_output = Perceptron([1, 1], -1)

def XNOR(x1, x2):

    h1 = xnor_and.predict([x1, x2])
    h2 = xnor_nor.predict([x1, x2])

    return xnor_output.predict([h1, h2])

print("XNOR:")
for x1, x2 in inputs:
    print(x1, x2, "->", XNOR(x1, x2))
```
