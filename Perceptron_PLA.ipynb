# Perceptron Learning Algorithm: All 16 Two-Input Boolean Functions

This notebook trains a **single perceptron using the Perceptron Learning Algorithm (PLA)** for all 16 two-input Boolean functions.

A single perceptron converges for the 14 linearly separable functions. **XOR and XNOR do not converge as single perceptrons**, so they are implemented using a small multilayer network whose individual perceptrons are trained using the same Perceptron Learning Algorithm.


```python
inputs = [
    [0, 0],
    [0, 1],
    [1, 0],
    [1, 1]
]

functions = {
    "FALSE":             [0, 0, 0, 0],
    "AND":               [0, 0, 0, 1],
    "x1 AND NOT x2":     [0, 0, 1, 0],
    "x1":                [0, 0, 1, 1],
    "NOT x1 AND x2":     [0, 1, 0, 0],
    "x2":                [0, 1, 0, 1],
    "XOR":               [0, 1, 1, 0],
    "OR":                [0, 1, 1, 1],
    "NOR":               [1, 0, 0, 0],
    "XNOR":              [1, 0, 0, 1],
    "NOT x2":            [1, 0, 1, 0],
    "x1 OR NOT x2":      [1, 0, 1, 1],
    "NOT x1":            [1, 1, 0, 0],
    "NOT x1 OR x2":      [1, 1, 0, 1],
    "NAND":              [1, 1, 1, 0],
    "TRUE":              [1, 1, 1, 1]
}
```

## 1. Perceptron with the Perceptron Learning Algorithm


```python
class Perceptron:
    def __init__(self, learning_rate=0.1, max_epochs=100):
        self.learning_rate = learning_rate
        self.max_epochs = max_epochs
        self.weights = None
        self.bias = 0.0
        self.epochs_used = 0
        self.converged = False

    def predict(self, x):
        z = sum(w * xi for w, xi in zip(self.weights, x)) + self.bias
        return 1 if z >= 0 else 0

    def fit(self, X, y):
        # Start from zero weights and bias
        self.weights = [0.0] * len(X[0])
        self.bias = 0.0
        self.converged = False

        for epoch in range(1, self.max_epochs + 1):
            errors = 0

            for x, target in zip(X, y):
                prediction = self.predict(x)
                error = target - prediction

                if error != 0:
                    errors += 1

                    # Perceptron Learning Algorithm : weight and bias update
                    for i in range(len(self.weights)):
                        self.weights[i] += self.learning_rate * error * x[i]

                    self.bias += self.learning_rate * error

            self.epochs_used = epoch

            if errors == 0:
                self.converged = True
                break

        return self
```

## 2. Train and test all 16 Boolean functions

For XOR and XNOR, a single perceptron will not converge because they are not linearly separable. The notebook explicitly shows this result.


```python
results = {}

for name, targets in functions.items():
    p = Perceptron(learning_rate=0.1, max_epochs=100)
    p.fit(inputs, targets)

    predictions = [p.predict(x) for x in inputs]

    results[name] = {
        "targets": targets,
        "predictions": predictions,
        "weights": p.weights.copy(),
        "bias": p.bias,
        "epochs": p.epochs_used,
        "converged": p.converged
    }

    print("=" * 60)
    print(f"Function: {name}")
    print(f"Target outputs : {targets}")
    print(f"Predictions    : {predictions}")
    print(f"Converged      : {p.converged}")
    print(f"Epochs used    : {p.epochs_used}")

    if p.converged:
        print(f"Weights        : {p.weights}")
        print(f"Bias           : {p.bias}")
    else:
        print("Result         : Cannot be implemented by a single perceptron")
```

## 3. Summary

Expected result:

- 14 functions converge using a single perceptron.
- XOR and XNOR do not converge using a single perceptron.


```python
print(f"{'Function':<22} {'Converged':<12} {'Output'}")
print("-" * 55)

for name, result in results.items():
    status = "Yes" if result["converged"] else "No"
    print(f"{name:<22} {status:<12} {result['predictions']}")
```

# XOR using Perceptron Learning Algorithm

We train three perceptrons:

1. Hidden perceptron 1 learns **OR**
2. Hidden perceptron 2 learns **NAND**
3. Output perceptron learns **AND** on the hidden outputs

Thus:

**XOR = AND(OR(x1, x2), NAND(x1, x2))**

Each individual perceptron is trained using the Perceptron Learning Algorithm.


```python
def train_gate(targets):
    p = Perceptron(learning_rate=0.1, max_epochs=100)
    p.fit(inputs, targets)

    if not p.converged:
        raise ValueError("This gate did not converge with a single perceptron.")

    return p


# Train hidden-layer gates using PLA
xor_or = train_gate(functions["OR"])
xor_nand = train_gate(functions["NAND"])

# Generate hidden-layer training data
xor_hidden_X = [
    [
        xor_or.predict(x),
        xor_nand.predict(x)
    ]
    for x in inputs
]

# XOR target
xor_targets = functions["XOR"]

# Train output perceptron using PLA
xor_output = Perceptron(learning_rate=0.1, max_epochs=100)
xor_output.fit(xor_hidden_X, xor_targets)


def XOR(x1, x2):
    h1 = xor_or.predict([x1, x2])
    h2 = xor_nand.predict([x1, x2])
    return xor_output.predict([h1, h2])


print("XOR RESULTS")
print("=" * 35)
print(f"OR converged      : {xor_or.converged}")
print(f"NAND converged    : {xor_nand.converged}")
print(f"Output converged  : {xor_output.converged}")
print()

for x, target in zip(inputs, xor_targets):
    print(f"{x} -> {XOR(x[0], x[1])}   (target = {target})")
```

# XNOR using Perceptron Learning Algorithm

We train three perceptrons:

1. Hidden perceptron 1 learns **AND**
2. Hidden perceptron 2 learns **NOR**
3. Output perceptron learns **OR** on the hidden outputs

Thus:

**XNOR = OR(AND(x1, x2), NOR(x1, x2))**

Again, every perceptron is trained using the Perceptron Learning Algorithm.


```python
# Train hidden-layer gates using PLA
xnor_and = train_gate(functions["AND"])
xnor_nor = train_gate(functions["NOR"])

# Generate hidden-layer training data
xnor_hidden_X = [
    [
        xnor_and.predict(x),
        xnor_nor.predict(x)
    ]
    for x in inputs
]

# XNOR target
xnor_targets = functions["XNOR"]

# Train output perceptron using PLA
xnor_output = Perceptron(learning_rate=0.1, max_epochs=100)
xnor_output.fit(xnor_hidden_X, xnor_targets)


def XNOR(x1, x2):
    h1 = xnor_and.predict([x1, x2])
    h2 = xnor_nor.predict([x1, x2])
    return xnor_output.predict([h1, h2])


print("XNOR RESULTS")
print("=" * 35)
print(f"AND converged     : {xnor_and.converged}")
print(f"NOR converged     : {xnor_nor.converged}")
print(f"Output converged  : {xnor_output.converged}")
print()

for x, target in zip(inputs, xnor_targets):
    print(f"{x} -> {XNOR(x[0], x[1])}   (target = {target})")
```

## Final conclusion

- **14 Boolean functions** are learned by a single perceptron using the Perceptron Learning Algorithm.
- **XOR and XNOR** are not learned by a single perceptron because they are not linearly separable.
- XOR and XNOR are implemented here using a **multilayer arrangement of perceptrons**, where every individual gate perceptron is trained using the Perceptron Learning Algorithm.
